## 4.1 Resampling to Balance Classes

We noted briefly that our classes were *unbalanced* which may be a problem. 

One way to deal with this is by **DOWNSAMPLING** the *majority* class (i.e., remove rows) or **UPSAMPLING** the *minority* class (i.e., duplicating rows). See other ways in {cite:t}`taskiran_comprehensive_2025`. I prefer the downsampling if anything!

### Your Turn: Downsample the Majority Class
:::{admonition} HANDS-ON
:class: red
- Split your train_df into `train_df_majority` and `..._minority` by filtering on the class = `0` (or e.g., `cohere`). Do the same for `val_df`
- Use the `resample` function from scikit-learn to DOWNSAMPLE the majority (see [docs](https://scikit-learn.org/stable/modules/generated/sklearn.utils.resample.html))
- Use the `pandas` function `pd.concat` to combine the majority and minority dfs for train (and repeat for val)
:::

> It's recommended to do resample after your split to avoid data leakage (duplicate rows in train and val/test, especially a problem for upsampling)
