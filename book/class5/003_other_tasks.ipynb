### 3.1 Transformers for Feature Extraction
As mentioned, during fine-tuning, all model parameters are trainable:
```{figure} ../figures/class5/chapter02_encoder-fine-tuning.png
---
name: fine-tuning
width: 60%
---
by {cite:t}`tunstall2022nlp`, chapter 2. Image downloaded through their [GitHub](https://github.com/nlp-with-transformers/notebooks/tree/main) (Apache-2.0 License).
```

However, **we can also keep the parameters frozen**, using the model outputs as features to train a lightweight neural network or non-neural method such as logistic regression (depending on our task):
```{figure} ../figures/class5/chapter02_encoder-feature-based.png
---
name: feature-based
width: 60%
---
by {cite:t}`tunstall2022nlp`, chapter 2. Image downloaded through their [GitHub](https://github.com/nlp-with-transformers/notebooks/tree/main) (Apache-2.0 License).
```
In practice, we extract the **last hidden states** for each text in our dataset (embeddings from final layer).

By using BERT as a feature extractor, we gain computational efficiency (since it does not require a GPU), and depending on the BERT model and task, it can perform nearly as well as full fine-tuning, making it a strong and practical baseline!
