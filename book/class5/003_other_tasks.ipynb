### 3.2 Named Entity Recognition
Another widely used task for BERT is Named Entity Recognition (NER), where BERT models are fine-tuned to identify different types of entities. 

These can range from simple categories such as PERSON, ORGANIZATION, and LOCATION to more fine-grained classifications such as the example below.
