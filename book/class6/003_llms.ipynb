## 3.1 RAGs 
LLM applications may require access to external knowledge beyond what was learned during training, especially when used in  industry contexts. **Retrieval-Augmented Generation** (RAG) addresses this by combining document retrieval with language generation:

```{figure} ../figures/class6/rag-flow-basic.png
---
name: basic-rag-flow
width: 100%
---
By [Langchain](https://blog.langchain.com/agentic-rag-with-langgraph/). Retrieval works by comparing the embedding of the user input (query) to document embeddings in a vector database, selecting relevant documents for the LLM to use as context.
```

:::{admonition} HANDS-ON
:class: red
Read through this [blogpost](https://huggingface.co/blog/ngxson/make-your-own-rag) on Hugging Face about RAGs. Feel free to try to implement it as well! 
:::

## 3.2 Fine-tuning Generative LLMs
Like we did with BERT, we can also fine-tune generative decoder-only LLMs:
:::{admonition} HANDS-ON
:class: red
[Read about Daniel's efforts](https://huggingface.co/blog/dvgodoy/fine-tuning-llm-hugging-face#fine-tuning-your-first-large-language-model-llm-with-pytorch-and-hugging-face) to fine-tune an LLM to write English like Yoda!

The tutorial is quite resource-heavy, so I recommend replacing the LLM with a smaller one - or just reading the tutorial rather than implementing it ;) 
:::

## 3.3 Can We Use Generative LLMs for Classification? 
Yes we can ... but they aren't always super good!
:::{admonition} HANDS-ON
:class: red
Read about [LLM classification performance](https://academic.oup.com/pnasnexus/article/4/4/pgaf069/8100281?login=false) in the paper by {cite:t}`kristensen-mclachlan_are_2025`. 

You can also look at how the paper implments few-shot prompting for LLM classification using the [stormtrooper](https://medium.com/@power.up1163/zero-and-few-shot-classification-with-transformers-llms-and-stormtrooper-bb2c6d139ef1) package by Márton Kardos (feel free to experiment with its implementation!)
:::
