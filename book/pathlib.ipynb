## Navigating from Scripts & Notebooks
Let's say we are working with both `clean_data.py` and `explore_data.ipynb` that are BOTH located in `src/data_processing/clean`:

To get the file path for `clean_data.py`, we would write: 
```python
path = Path(__file__)
print(path)
```
-> This prints `work/nlp/nlp-template/src/data_processing/clean/clean_data.py`

For the notebook `explore_data.ipynb`, we write: 
```python
path = Path.cwd()
print(path)
```
-> This prints `work/nlp/nlp-template/src/data_processing/clean`

:::{admonition} QUESTION
:class: red
Do you notice a difference between the two? 

<details>
<summary>ANSWER</summary>
While <code>Path(__file__)</code> for scripts writes the entire file path, <code>Path.cwd()</code> starts from the current working directory of the script <code>clean</code>. This changes how you should count!
<br><br>
This is of course a little confusing. However, it is most standard to use <code>Path(__file__)</code> for scripts, but it unfortunately does not work for notebook, hence <code>Path.cwd()</code>.
<br><br>
A good tip here is just to always print your <code>path = Path.cwd()</code> regardless of working with scripts or notebooks, and also print its parents e.g., <code>path.parents[1]</code> to understand where you land!
</details>
:::

### Counting Levels
As python is zero-indexed, we count each slash beginning from 0:
```
work/nlp/nlp-template/src/data_processing/clean/clean_data.py
         │            │   │               │       │
         │            │   │               │       └─ (the file)
         │            │   │               │
         │            │   │               └─ parents[0] (first parent)
         │            │   │
         │            │   └─ parents[1]
         │            │
         │            └─ parents[2]
         │
         └─ parents[3] ← To get to nlp-template
     
```

For the notebook, we need to be careful that we are starting somewhere else than the file!
```
work/nlp/nlp-template/src/data_processing/clean
         │            │   │               │
         │            │   │               └─ (current directory)
         │            │   │
         │            │   └─ parents[0] (first parent)
         │            │
         │            └─ parents[1]
         │
         └─ parents[2] ← To get to nlp-template
```
